```python
import numpy as np
import random
```

```python
def crea_tablero(lado = 10):
    tablero = np.full((lado,lado)," ")
    return tablero

def coloca_barco_plus(tablero, barco):
    # Nos devuelve el tablero si puede colocar el barco, si no devuelve False, y avise por pantalla
    tablero_temp = tablero.copy()
    num_max_filas = tablero.shape[0]
    num_max_columnas = tablero.shape[1]
    for pieza in barco:
        fila = pieza[0]
        columna = pieza[1]
        if fila < 0  or fila >= num_max_filas:
            print(f"No puedo poner la pieza {pieza} porque se sale del tablero")
            return False
        if columna <0 or columna>= num_max_columnas:
            print(f"No puedo poner la pieza {pieza} porque se sale del tablero")
            return False
        if tablero[pieza] == "O" or tablero[pieza] == "X":
            print(f"No puedo poner la pieza {pieza} porque hay otro barco")
            return False
        tablero_temp[pieza] = "O"
    return tablero_temp

def crea_barco_aleatorio(tablero,eslora = 4, num_intentos = 100):
    num_max_filas = tablero.shape[0]
    num_max_columnas = tablero.shape[1]
    while True:
        barco = []
        # Construimos el hipotetico barco
        pieza_original = (random.randint(0,num_max_filas-1),random.randint(0, num_max_columnas -1))
        print("Pieza original:", pieza_original)
        barco.append(pieza_original)
        orientacion = random.choice(["N","S","O","E"])
        print("Con orientacion", orientacion)
        fila = pieza_original[0]
        columna = pieza_original[1]
        for i in range(eslora -1):
            if orientacion == "N":
                fila -= 1
            elif orientacion  == "S":
                fila += 1
            elif orientacion == "E":
                columna += 1
            else:
                columna -= 1
            pieza = (fila,columna)
            barco.append(pieza)
        tablero_temp = coloca_barco_plus(tablero, barco)
        if type(tablero_temp) == np.ndarray:
            return tablero_temp
        print("Tengo que intentar colocar otro barco")

def recibir_disparo(tablero, coordenada):
    if tablero[coordenada] == "O":
        tablero[coordenada] = "X"
        print("Tocado")
        return "Tocado"
    elif tablero[coordenada] == "X":
        print("Agonia, deja de perder el tiempo, dispara a otro sitio")
        return "Repetido"
    else:
        tablero[coordenada] = "-"
        print("Agua")
        return "Agua"

def crea_barco_jugador(tablero,eslora = 4, num_intentos = 100):
    num_max_filas = tablero.shape[0]
    num_max_columnas = tablero.shape[1]
    while True:
        barco = []
        # Construimos el hipotetico barco
        x = int(input("Introduce la FILA: "))
        y = int(input("Introduce la COLUMNA"))
        pieza_original = (x,y)
        print("Pieza original:", pieza_original)
        barco.append(pieza_original)
        orientacion = input("Introduce la orientación del barco (N,S,E,O): ").upper()
            
        fila = pieza_original[0]
        columna = pieza_original[1]
        for i in range(eslora -1):
            if orientacion == "N":
                fila -= 1
            elif orientacion  == "S":
                fila += 1
            elif orientacion == "E":
                columna += 1
            else:
                columna -= 1
            pieza = (fila,columna)
            barco.append(pieza)
        tablero_temp = coloca_barco_plus(tablero, barco)
        print(tablero_temp)
        if type(tablero_temp) == np.ndarray:
            return tablero_temp
        print("Tengo que intentar colocar otro barco")
```

## **Definición de turnos para disparar**

Hay una funcion *recibir_disparo* que es la que se encarga de ejecutar los disparos. Pero necesitamos mas funciones entre medias, una para pedir las coordenadas, una para ver si hay barcos y otra que define los turnos. A su vez haremos una funcion de jugar que contendrá todas las anteriores (mencionando a turno, y dentro de turno se encuentran las otras)

```python
def recibir_disparo(tablero, coordenada):
    if tablero[coordenada] == "O":
        tablero[coordenada] = "X"
        print("Tocado")
        igualar_tableros(tablero, coordenada,cambio = "X")
        return "tocado"
    
    elif tablero[coordenada] in ("X", "-"):
        print("Agonia, deja de perder el tiempo, dispara a otro sitio")
        return "repetido"
    
    else:
        tablero[coordenada] = "-"
        print("Agua")
        igualar_tableros(tablero,coordenada, cambio = "-")
        return "agua"

def igualar_tableros(tablero, coordenada, cambio):
    if tablero is tb_j2:
        tablero = td_j1
    if tablero is tb_j1:
        tablero = td_j2

    if cambio == "X":
        tablero[coordenada] = "X"
    if cambio == "-":
        tablero[coordenada] = "-" 

def pide_coordenada(dim = 10):
    while True:
        fila = input("Fila (o escribe 'salir'): ")
        if fila.strip().lower() == "salir":
            return None
        columna = input("Columna (o escribe 'salir'): ")
        if columna.strip().lower() == "salir":
            return None
        try:
            fila = int(fila)
            columna = int(columna)
        except ValueError:
            print("Escribe números enteros")
            continue
        if 0 <= fila < dim and 0 <= columna < dim:
            return (fila, columna)
        print("Coordenada fuera del tablero")
```
